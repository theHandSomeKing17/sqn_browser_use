## 先看run
- 步数，每一步开始结束挂回调函数
- 拿到循环函数，并且用变量（初始值是None）来记录发生错误

```python
Agent.run()
│
├─ ① 准备运行环境
│    ├─ Browser Session
│    ├─ Skills
│    └─ Initial Actions
│
├─ ② 主循环
│
│    while steps <= max_steps
│          ↓
│      检查 pause
│          ↓
│      检查 failure
│          ↓
│      检查 stop
│          ↓
│      _execute_step()
│          ↓
│      is_done ?
│        ├─ yes → break
│        └─ no  → 下一轮
│
├─ ③ 返回 history
│
└─ ④ finally
     清理资源 / 日志 / event bus / close
```

## 重点看返回return history
**run的返回类型：AgentHistoryList[AgentStructuredOutput]**  
每一步的step包含多种信息，再把这些存入historylist，这样你就可以复盘啊，而不是只有一个最终结果！！！

```python
所以 history 不是单纯一个最终答案字符串。
它更像：
这一次 Agent 从开始到结束的完整执行记录。


AgentHistory(
    model_output=...,
    result=...,
    state=...,
    metadata=...,
)

一次 step
│
├── model_output
│    模型这一步输出了什么
│
├── result
│    Action 执行后发生了什么
│
├── state
│    当时浏览器是什么状态
│
└── metadata
     这一步相关的附加信息
```

## 小结

```python
run() 是 Agent 总控入口
运行前会准备 browser/session/skills/initial actions
进入 while 主循环
每轮先检查暂停、失败次数、停止条件、最大步数
真正的一步执行交给 _execute_step()
每一步结果会进入 history
history 用于最终结果、复盘、调试
最后做清理并返回 history
```
